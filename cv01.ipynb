**dtype**

In Python, dtype stands for data type and is used to specify the type of data in arrays and variables. This concept is especially important in libraries like NumPy and Pandas, as it allows you to specify the data type you use to store values, which can have a big impact on performance and memory usage. In Python, each variable or object has a specific data type (such as integer, decimal, string, etc.). In libraries like NumPy, which are designed for processing numerical and scientific data, you can specify a dtype for your arrays to define the data types exactly as you need.

**uint8**

An array of 8-bit unsigned integers (values ​​between 0 and 255).

**uint16**

uint16 means that each pixel of the image is stored as a 16-bit unsigned integer. In this case, each pixel can take on values ​​between 0 and 65535 (65536 different values).

**cmap**

In Python, and especially in graphics libraries like Matplotlib, a cmap or color map refers to a set of colors used to display images. If an image is grayscale, meaning the image has brightness values ​​that range from black to white, cmap can be used to convert these values ​​to different colors for better display or more detailed analysis of the image.

```python
ar1 = np.random.randint(0, 255, (20, 20), dtype = np.uint8)
ar1
```

```python
ar2 = np.random.randint(0, 65535, (20, 20), dtype = np.uint16)
ar2
```

**Convert Lenna color image to grayscale**

```python
Shape = lenna.shape
```

```python
if Shape[2] == 4:
  lenna = lenna[:, :, :3]

lenna_float = lenna.astype(float)
lenna_float.dtype
```

```python
 # Gray = 0.2989*R + 0.5870*G + 0.1140*B
lenna_gray = 0.2989 * lenna_float[:, :, 0] + 0.5870 * lenna_float[:, :, 1] + 0.1140 * lenna_float[:, :, 2]
lenna_gray.dtype
```

```python
lenna_gray
```

The resulting lenna_gray value is a float64 array whose values ​​are in the range 0 to 1 (i.e. 0 ≤ lenna_gray ≤ 1). These values ​​are not naturally suitable for representation as uint8, because uint8 is for values ​​between 0 and 255. To convert values ​​to uint8, we must multiply them by 255 so that the final values ​​are in the range 0 to 255, then convert to uint8.

```python
lenna_gray = np.clip(lenna_gray * 255, 0, 255)
lenna_gray8bit = lenna_gray.astype(np.uint8)
lenna_gray8bit.dtype
```

```python
lenna_eyes
```

```python
lenna_eyes_divided = lenna_eyes/10
lenna_eyes_divided
```

Showing changes in the eyes

```python
plt.figure(figsize = (12, 8))
plt.subplot(1, 2, 1)
plt.imshow(lenna_eyes, cmap = "grey")
plt.title("lenna eyes")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(lenna_eyes_divided, cmap = "grey")
plt.title("lenna eyes divided")
plt.axis("off")
plt.tight_layout()
plt.show()
```

The reason you don't see a difference between the two images in the plot is that the value lenna_eyes_divided = lenna_eyes / 10 is too small. In matplotlib, when you display an image with small values, imshow by default normalizes the values ​​between its minimum and maximum, so both images may look very similar. To see the difference clearly, use the vmin and vmax parameters to specify the range of the image display:

```python
plt.subplot(1, 2, 1)
plt.imshow(lenna_eyes, cmap='gray', vmin=0, vmax=255)
plt.title("lenna eyes")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(lenna_eyes_divided, cmap='gray', vmin=0, vmax=255)
plt.title("lenna eyes divided")
plt.axis("off")
plt.tight_layout()
plt.show()
```

Exercise 3
Constructing a color image from three separate channels

Building the basic array: Create three TWO-dimensional grayscale arrays
named R_arr, G_arr, and B_arr (with np.zeros or np.ones or filling with a constant value)

R_arr: Filled with the value 255 (full red)

G_arr: Filled with the value 100 (weak green)

B_arr: Filled with the value 0 (zero blue)
Resize: Resize all three arrays to
(e.g., by slicing)
Merge: Display: Display the final color image.

```python
# 1. ساخت آرایه‌های پایه 50x50
R_arr = np.full((50, 50), 255)  # قرمز کامل (255)
G_arr = np.full((50, 50), 100)  # سبز ضعیف (100)
B_arr = np.full((50, 50), 0)    # آبی صفر (0)

# 2. تغییر سایز آرایه‌ها به 100x100
R_arr_resized = R_arr.repeat(2, axis=0).repeat(2, axis=1)  # استفاده از repeat برای تغییر سایز
G_arr_resized = G_arr.repeat(2, axis=0).repeat(2, axis=1)
B_arr_resized = B_arr.repeat(2, axis=0).repeat(2, axis=1)

# 3. ترکیب سه کانال رنگی به تصویر رنگی
image = np.stack((R_arr_resized, G_arr_resized, B_arr_resized), axis=-1)

# 4. نمایش تصویر رنگی
plt.imshow(image)
plt.axis("off")
plt.title("Colored Image from R, G, B Channels")
plt.show()
```
